# Space Pizza MadLib

Enter five words to create a short space adventure.

## Instructions

Enter five words when prompted. The program will use them to create a short story.

```python
name = input("Give me a name: ")
adjective = input("Give me an adjective: ")
planet = input("Give me a planet: ")
food = input("Give me a food: ")
animal = input("Give me an animal: ")

print()
print(f"One day, {name} traveled to {planet} in a {adjective} spaceship.")
print(f"On the way, they shared {food} with a friendly {animal}.")
print("It was an unforgettable adventure!")
```
